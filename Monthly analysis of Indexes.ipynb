```python
import pandas as pd
import numpy as np
import yfinance as yf
import plotly.offline as pyo
import plotly.graph_objects as go

from plotly.subplots import make_subplots

pyo.init_notebook_mode(connected=True)
pd.options.plotting.backend = 'plotly'
```

```python
df = yf.download(['^NSEBANK', '^CNXIT', '^CNXPHARMA','^CNXFMCG', '^CNXAUTO', '^CNXMETAL'], start='2000-01-01')
df.dropna()
```

```python
df.Close.dropna()
```

# Log Returns

```python
log_returns = np.log(df.Close/df.Close.shift(1)).dropna()
log_returns
```

# Daily Cummulative Returns

```python
daily_cum_returns = np.exp(log_returns.cumsum())
daily_cum_returns.plot(title='Wealth Indices constructed from Cumulative Log Returns')
```

# Monthly Returns

```python
monthly_returns = (log_returns.resample('ME').sum())
monthly_returns
```

# Monthly Cummulative Returns

```python
monthly_cum_returns = np.exp(monthly_returns.cumsum())
monthly_cum_returns.plot()
```

```python
fig = make_subplots(rows=3, cols=2)

img0 = go.Histogram(x=log_returns['^CNXAUTO'], name='NIFTY AUTO')
img1 = go.Histogram(x=log_returns['^CNXFMCG'], name='NIFTY FMCG')
img2 = go.Histogram(x=log_returns['^CNXIT'], name='NIFTY IT')
img3 = go.Histogram(x=log_returns['^CNXMETAL'], name='NIFTY METAL')
img4 = go.Histogram(x=log_returns['^CNXPHARMA'], name='NIFTY PHARMA')
img5 = go.Histogram(x=log_returns['^NSEBANK'], name='NIFTY BANK')

fig.append_trace(img0, 1, 1)  #This is in terms of rows and columns
fig.append_trace(img1, 1, 2)
fig.append_trace(img2, 2, 1)
fig.append_trace(img3, 2, 2)
fig.append_trace(img4, 3, 1)
fig.append_trace(img5, 3, 2)

fig.update_layout(autosize=False, width=1100, height=600, title='Frequency of log returns')

fig.show()
```

```python
log_returns.plot(kind='box')
```

# Monthly Standard Deviation

```python
monthly_std = monthly_returns.std()
monthly_std
```

# Annualized Volatility

```python
annualized_volatility = monthly_std * np.sqrt(12)
annualized_volatility
```

# Monthly Historical Volatility

```python
Trading_days = 12

historical_volatility = monthly_returns.rolling(window=Trading_days).std() * np.sqrt(Trading_days)
historical_volatility.plot()
```

# Sharpe Ratio

```python
Rf = 0.06/12

excess_returns = monthly_returns - Rf
rolling_mean = excess_returns.rolling(window=Trading_days).mean()
rolling_vol = monthly_returns.rolling(window=Trading_days).std()

rolling_vol = rolling_vol.clip(lower=1e-1)

Sharpe_ratio = (rolling_mean / rolling_vol) * np.sqrt(12)
Sharpe_ratio.plot()
```
