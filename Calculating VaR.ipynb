```python
import pandas as pd
import yfinance as yf
import plotly.offline as pyo
import numpy as np

pyo.init_notebook_mode(connected=True)
pd.options.plotting.backend = 'plotly'
```

```python
df = yf.download(['RELIANCE.NS', 'TCS.NS', 'ONGC.NS'], start='2020-01-01', end='2025-12-31')['Close']
df
```

```python
log_returns = np.log(df/df.shift(1)).dropna()
log_returns
```

```python
log_returns.std()
```

```python
Position = df.iloc[-1]
Position
```

```python
from scipy.stats import norm
```

# This is for quantity

```python
shares = pd.Series({
    'ONGC.NS': 100,
    'RELIANCE.NS': 50,
    'TCS.NS': 10
})

position = shares * df.iloc[-1]        
sigma = log_returns.std()
z = abs(norm.ppf(0.95))

VaR_array = position * sigma * z
VaR_array
```

# This is for equaly capital allocation

```python
capital_per_stock = 100_000  # ₹1 lakh each

position_value = pd.Series(
    capital_per_stock,
    index=log_returns.columns
)

VaR_array2 = position_value * sigma * z
VaR_array2
```
